```python
# vocab = {
#     "collaboration": "協作",
#     "repository": "儲存庫",
#     "commit": "提交",
# }

# for en, zh in vocab.items():
#     print(f"{en}：{zh}")

from IPython.display import HTML, display

display(HTML("""
<!DOCTYPE html>
<html lang="zh-Hant">
<head>
<meta charset="UTF-8">

<style>
  .vocabuddy {
    font-family: Arial, sans-serif;
    text-align: center;
    background: #f4f6f5;
    padding: 30px;
    border-radius: 16px;
  }

  .vocabuddy h1 {
    color: #0f766e;
  }

  #card {
    max-width: 340px;
    margin: 30px auto;
    padding: 40px 24px;
    background: white;
    border-radius: 14px;
    box-shadow: 0 4px 16px rgba(0,0,0,.08);
    font-size: 2rem;
    cursor: pointer;
    user-select: none;
  }

  #hint {
    color: #888;
  }

  .vocabuddy button {
    margin: 8px;
    padding: 10px 22px;
    font-size: 1rem;
    border: none;
    border-radius: 8px;
    background: #0f766e;
    color: white;
    cursor: pointer;
  }
</style>

<div class="vocabuddy">

  <h1>VocaBuddy 背單字夥伴</h1>

  <div id="card" onclick="flipCard()">algorithm</div>

  <p id="hint">
    點卡片看中文，點「下一張」換單字
  </p>

  <button onclick="nextWord()">下一張</button>
  <button onclick="randomWord()">隨機單字</button>

</div>

<script>

const WORDS = [
  ["algorithm","演算法"],
  ["recursion","遞迴"],
  ["vocabulary","字彙"],
  ["memorize","背誦"],
  ["efficient","有效率的"],
  ["persist","堅持；持續"],
  ["retrieve","取回；檢索"],
  ["interval","間隔"],

  // 新增單字
  ["collaboration","協作"],
  ["repository","儲存庫"],
  ["commit","提交"]
];

let i = 0;
let flipped = false;

const card = document.getElementById("card");

// 翻面
function flipCard() {
  flipped = !flipped;
  render();
}

// 下一張
function nextWord() {
  i = (i + 1) % WORDS.length;
  flipped = false;
  render();
}

// 隨機抽取單字
function randomWord() {
  i = Math.floor(Math.random() * WORDS.length);
  flipped = false;
  render();
}

// 更新卡片
function render() {
  card.textContent = WORDS[i][flipped ? 1 : 0];
}

</script>
"""))
```
